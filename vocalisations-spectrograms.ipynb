```python
import os
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
# Create the output directory if it doesn't exist
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Function to create and save spectrograms
def generate_spectrograms(audio_segments_dir, output_dir):
    for filename in sorted(os.listdir(audio_segments_dir)):
        if filename.endswith(".wav"):  # Process only .wav files
            file_path = os.path.join(audio_segments_dir, filename)
            print(f"Processing {filename}...")

            # Load the audio file
            y, sr = librosa.load(file_path, sr=None)  # Load with original sampling rate

            # Compute the spectrogram (Short-Time Fourier Transform - STFT)
            S = librosa.stft(y)
            S_db = librosa.amplitude_to_db(np.abs(S), ref=np.max)

            # Plot the spectrogram
            plt.figure(figsize=(10, 4))
            librosa.display.specshow(S_db, sr=sr, x_axis='time', y_axis='hz', cmap='viridis')
            plt.colorbar(format='%+2.0f dB')
            plt.title(f"Spectrogram of {filename}")
            plt.tight_layout()

            # Save the spectrogram as an image
            output_file = os.path.join(output_dir, f"{os.path.splitext(filename)[0]}_spectrogram.png")
            plt.savefig(output_file)
            plt.close()

            print(f"Spectrogram saved as: {output_file}")

generate_spectrograms(audio_segments_dir, output_dir)

print("Spectrogram generation complete!")
```

```python
# Directory containing spectrogram images
spectrogram_dir = output_dir +"/spectrograms" 
cropped_dir = spectrogram_dir+"/cropped_spectrograms" 

# good enough but there is a white stripe on the right side
desired_width = 662
desired_height = 285

# Create the output directory if it doesn't exist
if not os.path.exists(cropped_dir):
    os.makedirs(cropped_dir)

# Function to crop all images to the desired size
def crop_to_fixed_size(input_dir, output_dir, width, height):
    for filename in sorted(os.listdir(input_dir)):
        if filename.endswith(".png"):  # Process only .png files
            file_path = os.path.join(input_dir, filename)
            print(f"Cropping {filename} to {width}x{height}...")

            # Open the image
            with Image.open(file_path) as img:
                # Ensure the image is large enough
                img_width, img_height = img.size
                if img_width < width or img_height < height:
                    print(f"Warning: {filename} is smaller than {width}x{height}. Skipping.")
                    continue

                # Calculate cropping box for the center
                left = (img_width - width) // 2
                top = (img_height - height) // 2
                right = left + width
                bottom = top + height

                # Crop the image
                cropped_img = img.crop((left, top, right, bottom))

                # Save the cropped image
                cropped_img_path = os.path.join(output_dir, filename)
                cropped_img.save(cropped_img_path)
                print(f"Saved cropped image: {cropped_img_path}")

crop_to_fixed_size(spectrogram_dir, cropped_dir, desired_width, desired_height)

print("Cropping complete!")
```
