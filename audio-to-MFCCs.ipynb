```python
import os
import librosa
import numpy as np
import pandas as pd
import re
```

```python
def extract_number(file_name):
    match = re.search(r'\d+', file_name)
    return int(match.group()) if match else float('inf')

def extract_mfccs_from_folder(folder_path, n_mfcc=13, sr=22050, max_length=100):
    data = []
    
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".wav"):
            file_path = os.path.join(folder_path, file_name)
            audio, _ = librosa.load(file_path, sr=sr)
            mfccs = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=n_mfcc)
            
            # Padding or truncating to ensure equal length
            if mfccs.shape[1] < max_length:
                pad_width = max_length - mfccs.shape[1]
                mfccs = np.pad(mfccs, ((0, 0), (0, pad_width)), mode='constant')
            else:
                mfccs = mfccs[:, :max_length]
            
            data.append([file_name, mfccs])
    
    df = pd.DataFrame(data, columns=["file_name", "mfccs"])
    df["sort_key"] = df["file_name"].apply(extract_number)
    df = df.sort_values(by="sort_key").drop(columns=["sort_key"]).reset_index(drop=True)
    return df


folder_path = #adapt path
df = extract_mfccs_from_folder(folder_path)
print(df.head())
```
