## Audio to Raw MFCCs

```python
import os
import librosa
import numpy as np
import pandas as pd
import re
```

```python
# Flexible lenght in the MFCCs (no padding or truncation) and we will extract the mean and std of each MFCC coefficient across time as our baseline features. This results in 40 features (20 means + 20 stds) per audio file.

np.set_printoptions(threshold=np.inf)

def extract_number(file_name):
    match = re.search(r'\d+', file_name)
    return int(match.group()) if match else float('inf')

def create_raw_mfcc_csv(folder_path, output_csv, n_mfcc=20, sr=22050):
    data = []
    
    all_files = [f for f in os.listdir(folder_path) if f.endswith(".wav")]
    all_files.sort(key=extract_number)
    
    print(f"Found {len(all_files)} files. Starting extraction...")

    for file_name in all_files:
        file_path = os.path.join(folder_path, file_name)
        
        audio, _ = librosa.load(file_path, sr=sr)
        
        mfccs = librosa.feature.mfcc(y=audio, sr=sr, n_mfcc=n_mfcc)
        
        mfcc_string = np.array2string(mfccs, separator=',', precision=10)
        
        data.append([file_name, mfcc_string])
    
    df = pd.DataFrame(data, columns=["file_name", "mfccs"])
    df.to_csv(output_csv, index=False)
    print(f"Success! Full MFCC data saved to: {output_csv}")

# --- EXECUTION ---
folder_path = r'C:/Users/nphan/Desktop/UZH/uzh II/Advanced Machine Learning/denoised_audios' 
output_name = r'C:/Users/nphan/Desktop/UZH/uzh II/Advanced Machine Learning/MFCC_denoised_audios.csv'

create_raw_mfcc_csv(folder_path, output_name)
```
